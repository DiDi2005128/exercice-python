```python
import seaborn as sns
import pandas as pd

taxis = sns.load_dataset("taxis")

print(taxis.shape)
```

Nombre de course et chiffre d'affaire par jour

```python
taxis["jour"] = taxis["pickup"].dt.date

par_jour = taxis.groupby("jour").agg(
    nombre_courses=("pickup", "count"),
    chiffre_affaires=("total", "sum")
)

print(par_jour)
```

L'heure de la journée la plus chargé

```python
taxis["heure"] = taxis["pickup"].dt.hour

courses_par_heure = taxis.groupby("heure").size()

print(courses_par_heure.sort_values(ascending=False))
```

Le jour de la semaine avec le pourboire moyen le plus élevé

```python
taxis["jour_semaine"] = taxis["pickup"].dt.day_name()

pourboires = taxis.groupby("jour_semaine")["tip"].mean()

print(pourboires.sort_values(ascending=False))
```

La durée de chaque course en minutes, repérerage et traitement des
durées aberrantes

```python
taxis["duree_minutes"] = (
    taxis["dropoff"] - taxis["pickup"]
).dt.total_seconds() / 60

print(taxis["duree_minutes"].describe())

print(taxis[
    (taxis["duree_minutes"] < 0) |
    (taxis["duree_minutes"] > 180)
][["pickup", "dropoff", "duree_minutes"]])
```
