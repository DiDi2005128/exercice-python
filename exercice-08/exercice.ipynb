```python
import seaborn as sns
import pandas as pd
titanic = sns.load_dataset("titanic")
```

Traiter les manquants

```python
#Remplacer les âges manquants par la médiane de chaque groupe
titanic["age"] = titanic.groupby( 
    ["pclass", "sex", "embark_town", "deck"], 
    dropna=False 
    )["age"].transform( 
        lambda x: x.fillna(x.median()) 
    )

#Remplacer les âges encore manquants par la médiane globale
titanic["age"] = titanic["age"].fillna(titanic["age"].median())

#Vérifier le nombre de valeurs manquantes restantes
print("Valeurs manquantes dans age :", titanic["age"].isnull().sum())
```

examiner les doublons 

```python
print(titanic.duplicated().sum())
print(titanic[titanic.duplicated()])
print(titanic[titanic.duplicated(keep=False)])
```

supprimer les doublons 

```python
titanic = titanic.drop_duplicates()
print("nombre de doublons restants: ", titanic.duplicated().sum())
```

supprimer les colonnes redondantes

```python
titanic = titanic.drop(columns=["embark_town"])
```

corriger les types de données

```python
titanic["survived"] = titanic["survived"].astype("int64") 
titanic["pclass"] = titanic["pclass"].astype("int64") 
titanic["sex"] = titanic["sex"].astype("category") 
titanic["embarked"] = titanic["embarked"].astype("category") 
titanic["who"] = titanic["who"].astype("category")
```

```python
print(titanic.dtypes)
print(titanic.shape)
```

reperer les billets hors bornes

```python
#Calcul des quartiles
Q1 = titanic["fare"].quantile(0.25) 
Q3 = titanic["fare"].quantile(0.75)

#Calcul de l'intervalle interquartile
IQR = Q3 - Q1

#Définir les bornes
borne_inf = Q1 - 1.5 * IQR 
borne_sup = Q3 + 1.5 * IQR

#Repérer les billets hors bornes
billets_hors_bornes = titanic[ (titanic["fare"] < borne_inf) | (titanic["fare"] > borne_sup) ]
print(billets_hors_bornes[["fare", "pclass", "embarked"]])
```
