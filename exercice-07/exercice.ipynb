```python
import seaborn as sns
import pandas as pd
titanic = sns.load_dataset("titanic")
```

```python
#passager de moin de 18 ans
print((titanic["age"] < 18 ).sum())
```

```python
#les 5 billets les plus cheres
print(titanic.nlargest(5, "fare")[["fare", "pclass", "embark_town"]])
```

```python
#femme de 1er classe
premiere_classe = titanic[titanic["pclass"] == 1]

part_femmes = (premiere_classe["sex"] == "female").mean() * 100

print("Part des femmes :", part_femmes, "%")
```

```python
#Femmes de 1re classe
femmes_1 = titanic[ (titanic["pclass"] == 1) & (titanic["sex"] == "female") ]

#Hommes de 3e classe
hommes_3 = titanic[ (titanic["pclass"] == 3) & (titanic["sex"] == "male") ]

#Calcul des taux de survie
print("Femmes de 1re classe :", femmes_1["survived"].mean() * 100, "%") 
print("Hommes de 3e classe :", hommes_3["survived"].mean() * 100, "%")
```
