```python
import pandas as pd
import seaborn as sns
import matplotlib.pyplot as plt
titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")
taxis = sns.load_dataset("taxis")
```

L'âge qu'avaient les passagers

```python
titanic["age"].hist(bins=20)
```

les passages avaient en général entre 30 et 40 ans 

La survie selon la classe et le sexe

```python
sns.barplot(
    data=titanic,
    x="pclass",
    y="survived",
    hue="sex"
)

plt.title("Taux de survie selon la classe et le sexe")
plt.xlabel("Classe du passager")
plt.ylabel("Taux de survie")
plt.legend(title="Sexe")
plt.show()
```

le prix du billets selon la classe

```python
sns.boxplot(data=titanic, x="pclass", y="fare")
plt.title("Prix du billet selon la classe")
plt.xlabel("Classe")
plt.ylabel("Prix du billet")
plt.show()
```

Le chiffre d'affaire jour après jour

```python
taxis["pickup"] = pd.to_datetime(taxis["pickup"])

taxis["jour"] = taxis["pickup"].dt.date

ca_journalier = taxis.groupby("jour")["total"].sum()

ca_journalier.plot(kind="line", figsize=(12, 5))

plt.title("Chiffre d'affaires des taxis jour après jour")
plt.xlabel("Jour")
plt.ylabel("Chiffre d'affaires ($)")
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

Le lien entre distance et prix

```python
plt.scatter(taxis["distance"], taxis["fare"])

plt.title("Lien entre la distance et le prix")
plt.xlabel("Distance parcourue")
plt.ylabel("Prix de la course ($)")

plt.show()
```
