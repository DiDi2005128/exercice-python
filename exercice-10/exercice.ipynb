```python
import pandas as pd

titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")
```

creer la table ports

```python
ports = pd.DataFrame({
    "embarked": ["S", "C", "Q"],
    "pays": ["Angleterre", "France", "Irlande"],
}) 
```

joindre la table a titanic_propre

```python
noms_ports = {
    "S": "Southampton",
    "C": "Cherbourg",
    "Q": "Queenstown"
}

titanic["embark_town"] = titanic["embarked"].map(noms_ports)
print(titanic[["embarked", "embark_town"]].head())
```

```python
nb_avant = len(titanic)

titanic_propre = titanic.merge(
    ports,
    on="embarked",
    how="left"
)

print("Lignes avant :", nb_avant)
print("Lignes après :", len(titanic_propre))
print(titanic_propre[["embarked", "embark_town", "pays"]].head())
```

compter le nombre de passager

```python
resultat = titanic_propre.groupby("pays").agg(
    nombre_passagers=("survived", "count"),
    taux_survie=("survived", "mean")
)

resultat["taux_survie"] = resultat["taux_survie"] * 100

print(resultat.round(2))
```

retirer Queenstown de ports

```python
ports = ports[ports["embarked"] != "Q"]

print(ports)
```

refaire la jointure avec inner

```python
titanic_inner = titanic.merge(
    ports,
    on="embarked",
    how="inner"
)

print("Nombre de lignes avec inner :", len(titanic_inner))
```

refaire la jointure avec left

```python
titanic_left = titanic.merge(
    ports,
    on="embarked",
    how="left"
)

print("Nombre de lignes avec left :", len(titanic_left))
print(titanic_left[["embarked", "pays"]].tail(10))
```

comparer les jointures

```python
print("Lignes dans titanic :", len(titanic))
print("Lignes après inner :", len(titanic_inner))
print("Lignes après left :", len(titanic_left))
```

Avec inner le nombre de ligne diminue mais avec left le nombre de ligne reste intacte
