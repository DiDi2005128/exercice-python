Determiner si les femmes ont mieux survécu que les hommes

```python
import pandas as pd

titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")

taux_survie = titanic.groupby("sex")["survived"].mean() * 100

print(taux_survie)
```

Le taux de survie est plus eleve chez les femmes donc elles ont plus survecu

Taux de survie par tranche d'age

```python
import pandas as pd

titanic = pd.read_csv("../exercice-08/data/titanic_propre.csv")

#Définir les tranches d'âge
bornes = [0, 12, 18, 35, 60, float("inf")] 
labels = ["Enfants", "Adolescents", "Adultes 18-34", "Adultes 35-59", "Seniors 60+"]
titanic["tranche_age"] = pd.cut( titanic["age"], bins=bornes, labels=labels, right=False )

#Calculer les effectifs et les taux de survie
resultats = titanic.groupby( "tranche_age", observed=False ).agg( nombre_passagers=("survived", "count"), nombre_survivants=("survived", "sum"), taux_survie=("survived", "mean") )
resultats["taux_survie"] *= 100
print(resultats.round(2).sort_values("taux_survie", ascending=False))
```

En se basant juste sur le taux de survie par rapport a la tranche d'age on peut penser que les adolescents ont ete priviligié

Savoir quel port a le prix moyen le plus élevé

```python
#Calculer le prix moyen des billets par port
prix_moyen = titanic.groupby("embarked")["fare"].mean()

#Afficher les ports du plus cher au moins cher
print(prix_moyen.sort_values(ascending=False).round(2))
```
